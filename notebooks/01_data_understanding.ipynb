## 8. Define Decisioned Applications

For approval-rate analysis, restrict the population to applications with a clear lending decision:

- **Approved:** Loan originated (`1`) or approved but not accepted (`2`)
- **Denied:** Application denied (`3`)

Withdrawn, incomplete, purchased, and preapproval records are excluded because they do not represent comparable final underwriting decisions in this analysis.

```python
decisioned_df = df[
  df["action_taken"].isin([1, 2, 3])
].copy()
```

```python
decisioned_df.shape
```

```python
decisioned_df["approved"] = (
    decisioned_df["action_taken"].isin([1, 2])
).astype(int)
```

```python
decisioned_df[
    ["action_taken", "action_label", "approved"]
].head(10)
```

## 9. Approval Rates by Loan Type

Compare approval outcomes across major mortgage loan types within the decisioned application population. Both application volume and approval rate are reported because percentage differences should be interpreted in the context of sample size.

```python
overall_approval_rate = decisioned_df["approved"].mean()

print(f"Overall approval rate: {overall_approval_rate:.2%}")
```

```python
loan_type_labels = {
    1: "Conventional",
    2: "FHA",
    3: "VA",
    4: "USDA/RHS"
}

decisioned_df["loan_type_label"] = (
    decisioned_df["loan_type"].map(loan_type_labels)
)
```

## Loan Type Approval Summary

```python
loan_type_summary = (
    decisioned_df
    .groupby("loan_type_label")["approved"]
    .agg(["count", "sum", "mean"])
)
```

```python
loan_type_summary = loan_type_summary.rename(
    columns={
        "count": "applications",
        "sum": "approved",
        "mean": "approval_rate"
    }
)
```

```python
loan_type_summary["denied"] = (
    loan_type_summary["applications"]
    - loan_type_summary["approved"]
)
```

```python
loan_type_summary["approval_rate_pct"] = (
    loan_type_summary["approval_rate"] * 100
)

loan_type_summary = loan_type_summary.sort_values(
    "approval_rate_pct",
    ascending=False
)

loan_type_summary[
    ["applications", "approved", "denied", "approval_rate_pct"]
].round(2)
```

### Validation

```python
assert (
    loan_type_summary["approved"]
    + loan_type_summary["denied"]
    == loan_type_summary["applications"]
).all()
```

### Interpretation

Among decisioned applications, the overall approval rate is approximately **75.2%**. Approval rates vary across loan types, ranging from **74.3% for conventional loans** to **82.8% for VA loans**.

These descriptive differences should not be interpreted as evidence that one loan type is inherently easier to obtain. The borrower populations and application characteristics differ across loan programs, including financial profiles, loan amounts, property characteristics, and loan purposes. VA loans also serve an eligibility-restricted population, while the USDA/RHS group contains a much smaller sample than the other loan types.

Therefore, these results show an **association between loan type and observed approval outcomes**, not a causal effect of loan type on approval probability.

## Loan Purpose Code Mapping

Map HMDA loan-purpose codes to descriptive labels before performing group-level analysis.

```python
loan_purpose_labels = {
    1: "Home Purchase",
    2: "Home Improvement",
    4: "Other Purpose",
    5: "Not Applicable",
    31: "Refinancing",
    32: "Cash-out Refinancing"
}
```

```python
decisioned_df["loan_purpose_label"] = (
    decisioned_df["loan_purpose"].map(loan_purpose_labels)
)
```

```python
decisioned_df[
    ["loan_purpose", "loan_purpose_label"]
].drop_duplicates().sort_values("loan_purpose")
```

## Loan Purpose Approval Summary

Summarize application volume, approvals, denials, and approval rates for each loan-purpose category.

```python
loan_purpose_summary = (
    decisioned_df
    .groupby("loan_purpose_label")["approved"]
    .agg(["count", "sum", "mean"])
)
```

```python
loan_purpose_summary = loan_purpose_summary.rename(
    columns={
        "count": "applications",
        "sum": "approved",
        "mean": "approval_rate"
    }
)
```

```python
loan_purpose_summary["denied"] = (
    loan_purpose_summary["applications"]
    - loan_purpose_summary["approved"]
)
```

```python
loan_purpose_summary["approval_rate_pct"] = (
    loan_purpose_summary["approval_rate"] * 100
)
```

```python
loan_purpose_summary = loan_purpose_summary.sort_values(
    "approval_rate_pct",
    ascending=False
)
```

```python
loan_purpose_summary[
    [
        "applications",
        "approved",
        "denied",
        "approval_rate_pct"
    ]
].round(2)
```

### Validation

Confirm that approved and denied applications reconcile to the total decisioned applications within each loan-purpose category.

```python
assert (
    loan_purpose_summary["approved"]
    + loan_purpose_summary["denied"]
    == loan_purpose_summary["applications"]
).all()
```

## Interpretation

Among loan purposes with clearly defined categories, **Home Purchase** has the highest observed approval rate at approximately **88.43%**, while **Other Purpose** has the lowest observed approval rate at approximately **55.55%**.

Traditional refinancing applications show an observed approval rate of approximately **76.26%**, compared with **69.74%** for Cash-out Refinancing.

The **Not Applicable** category has the highest numerical approval rate at approximately **89.19%**, but it contains only **509 decisioned applications**, substantially fewer than the other categories. Its approval rate should therefore be interpreted cautiously.

These descriptive differences should not be interpreted as evidence that one loan purpose is inherently easier to obtain approval for than another. Borrower and application characteristics may differ systematically across loan-purpose groups, including income, debt-to-income ratio, loan-to-value ratio, loan amount, property characteristics, and loan type.

Therefore, the results indicate an **association between loan purpose and observed approval outcomes**, rather than a causal effect of loan purpose on approval probability.
